#### Conductor y autobus

```python
# --- SIMULADOR DE ASISTENTE LÓGICO PARA AUTOBUSES ---

# Inicialización de variables (Estado inicial del autobús)
combustible = 50.0  # Porcentaje de combustible
pasajeros = 0
capacidad_maxima = 30
parada_actual = 1
estado_sistema = "Activo"

print("--- Sistema de Inteligencia Lógica del Conductor ---")

# Ciclo principal interactivo (Mantiene el programa funcionando) [1, 4]
while estado_sistema == "Activo":
    print(f"\nESTADO ACTUAL: Parada {parada_actual} | Pasajeros: {pasajeros} | Combustible: {combustible}%")
    print("1. Abrir puertas (Recoger pasajeros)")
    print("2. Consultar a la IA (¿Qué debo hacer?)")
    print("3. Avanzar a la siguiente parada")
    print("4. Ir a la gasolinera")
    print("5. Apagar sistema (Salir)")

    opcion = input("\nSeleccione una opción (1-5): ")

    # Estructura condicional para el menú [2, 5]
    if opcion == "1":
        # Operador aritmético para sumar pasajeros [6]
        nuevos = int(input("¿Cuántos pasajeros suben?: "))
        if (pasajeros + nuevos) <= capacidad_maxima:
            pasajeros = pasajeros + nuevos
            print(f"Éxito: Ahora hay {pasajeros} pasajeros a bordo.")
        else:
            # Operador relacional para comparar capacidad [7]
            print("Error: No hay espacio suficiente para todos.")

    elif opcion == "2":
        # --- SIMULACIÓN DE INTELIGENCIA ARTIFICIAL ---
        # La IA analiza múltiples condiciones usando operadores lógicos (and, or) [8, 9]
        print("\n[IA ANALIZANDO...]")

        if combustible < 15:
            print("DECISIÓN: ¡Alerta! El combustible es crítico. Vaya a la gasolinera de inmediato.")

        elif pasajeros >= capacidad_maxima:
            print("DECISIÓN: Autobús lleno. No recoja más pasajeros en la siguiente parada.")

        elif combustible < 30 and pasajeros > (capacidad_maxima / 2):
            # Combinación de operadores aritméticos y relacionales [10]
            print("DECISIÓN: Peso alto y poco combustible. Conduzca despacio hacia la estación.")

        else:
            print("DECISIÓN: Todo normal. Continúe la ruta de forma estándar.")

    elif opcion == "3":
        # Simulación de avance y consumo de combustible [6, 11]
        if combustible > 5:
            parada_actual = parada_actual + 1
            combustible = combustible - 10.5
            print("Avanzando... El autobús ha llegado a la siguiente parada.")
        else:
            print("Error: Sin combustible suficiente para avanzar.")

    elif opcion == "4":
        # Recarga de combustible
        combustible = 100.0
        print("Depósito lleno. ¡Listo para continuar!")

    elif opcion == "5":
        # Finalización del ciclo while [4, 12]
        estado_sistema = "Apagado"
        print("Cerrando sesión del Asistente Lógico. ¡Buen viaje!")

    else:
        print("Opción no válida. Intente de nuevo.")

print("\n--- PROGRAMA FINALIZADO ---")
print ("progama creado por nahum flores NL = 21")
```
